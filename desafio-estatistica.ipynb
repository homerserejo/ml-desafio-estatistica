### Desafio Técnico (Estatística para Devs)
Com base nas aulas do módulo de Estatística, utilize os métodos da biblioteca Pandas para tratar o dicionário abaixo.

Importando as dependências

```python
# Importando as dependências
import pandas as pd
import matplotlib.pyplot as plt
```

Declarando o dicionário

```python
# Dicionário de faturamento
dict_faturamento = {
    'data_ref': [
        '2023-01-01', 
        '2020-02-01', 
        '2021-03-01', 
        '2022-04-01', 
        '2023-05-01',
        '2023-06-01', 
        '2020-07-01', 
        '2021-08-01', 
        '2022-09-01', 
        '2023-10-01',
        '2022-11-01', 
        '2023-12-01',
        ],
    'valor': [
        400000, 
        890000, 
        760000, 
        430000, 
        920000,
        340000, 
        800000, 
        500000, 
        200000, 
        900000,
        570000, 
        995000,
        ]
}
```

```python
# Testando o dicionário
dict_faturamento
```

Gerando o DataFrame

```python
# Gerando o DataFrame
df_faturamento = pd.DataFrame.from_dict(dict_faturamento)
```

```python
# Testando o DataFrame
df_faturamento
```

```python
# Conferindo as classes do DF
df_faturamento.info()
```

```python
# Ajustando o Dtype de 'data_ref'

df_faturamento['data_ref'] = pd.to_datetime(df_faturamento['data_ref'])
```

```python
# Conferindo o ajuste
df_faturamento.info()
```

Trazer a média das vendas

```python
# Média
df_faturamento['valor'].mean()
```

```python
# Média
df_faturamento.valor.mean()
```

Criar um gráfico de barras vertical mostrando o mês de referência e o valor

```python
df_faturamento.sort_values('data_ref').plot.bar(x='data_ref', y='valor')
```

Criar um gráfico de linhas mostrando o mês de referência e o valor

```python
df_faturamento.sort_values('data_ref').plot.line(x='data_ref', y='valor')
```
